```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = sns.load_dataset('penguins')
df.head()
```

```python
# Task 1
ax = sns.barplot(data = df,x='species', y= 'body_mass_g',errorbar= None)
ax.bar_label(ax.containers[0],fontsize=10)
ax.set_xlabel("Species")
ax.set_ylabel("Body Mass (grams)")
ax.set_title("Average Body Mass By Species")
```

```python
# mean > median > mode right skewed
```

```python
# Task 2
ax = sns.countplot(data=df,x='species')
ax.set_ylabel('count')
ax.set_title('Species Distribution')

# Adelie is most common species
```

```python
# Task 3: Histogram: Distribution Shape
ax= sns.histplot(data=df,x='body_mass_g',kde=True)
ax.set_title('Distribution of Body Mass (Grams)')
ax.set_xlabel('Body Mass (Grams)')
print(f'mean: {df['body_mass_g'].mean()} \n median: {df['body_mass_g'].median()} \n mode: {df['body_mass_g']}')
# right skewed
```

```python
# Task 4 : Box Plot: Visual Outlier Detection
ax = sns.boxplot(data=df,x= 'species',y='body_mass_g')
ax.set_title('Distribution By Categories')
# 2 outliers in the chinsrap species
```

```python
# Task 5
q1= df['body_mass_g'].quantile(0.25)
q3 = df['body_mass_g'].quantile(0.75)
iqr = q3-q1
lower = q1-1.5*iqr
upper = q3+1.5*iqr
outliers = df[(df['body_mass_g'] < lower)|(df['body_mass_g'] > upper)]
print(df[(df['body_mass_g'] < lower)|(df['body_mass_g'] > upper)])

sns.boxplot(data=df,y='body_mass_g')

# task 4 and 5 resutls don't match because task 4 boxplots are based on each category while task 5
# boxplot take the whole column as it is
```

```python
# Task 6 : Outlier Detection: Z-score Comparison
from scipy import stats
df['zscore'] = stats.zscore(df['body_mass_g'])
outliers_z = df[df['zscore'].abs() > 3]
print(len(outliers_z))
# as our body_mass_g column is right skewed where as z score works best for normal distribution
```

```python
# Scatter Plot: Relationship Between Two Variables
ax = sns.scatterplot(data=df,x='body_mass_g',y= 'flipper_length_mm')
ax.set_title('Relationship b/w Body Mass and Flipper Lenght')
```
