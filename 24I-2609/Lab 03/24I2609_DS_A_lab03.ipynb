```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data.csv', encoding='latin1')
df.head()
```

```python
df.info()
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns
```

# Task 1

```python
summarized_monthly_sales = df.groupby(df['order_date'].dt.month)['sales'].sum().round(2)
```

```python
summarized_monthly_sales
```

```python
ax = sns.barplot(summarized_monthly_sales)
plt.title('Monthly Sales')
plt.xlabel('Month')
```

```python
plt.figure(figsize=(6,6))
plt.pie(summarized_monthly_sales,labels=summarized_monthly_sales.index,autopct='%1.1f%%')
```

```python
# seaborn line chart show all month numbers

ax= sns.lineplot(x=summarized_monthly_sales.index,y=summarized_monthly_sales.values)
ax.set_xticks(range(1,13))

month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
ax.set_xticklabels(month_names)
plt.title('Monthly Sales')
plt.ylabel('Sales')
plt.xlabel('')

```

```python
# pie chart is not good for showing trends whereas bar plot is a week choice for showing trends line plot display the trends perfectly
```

# Task 2

```python
top_five_sub_cat_sales = df.groupby('sub-category')['sales'].sum().sort_values(ascending=False).head(5).round(2)
sns.barplot(top_five_sub_cat_sales)
plt.title('Top Five Sub-Category Sales')
plt.xlabel('Sub-Category')
plt.ylabel('Sales')
```

```python
# Bar plot is good for comparing categorical data whereas pie chart might not show comparision for subcategories having almost same sales
```

```python
ax= sns.lineplot(x=summarized_monthly_sales.index,y=summarized_monthly_sales.values)
ax.set_xticks(range(1,13))

month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
ax.set_xticklabels(month_names)
plt.title('Monthly Sales')
plt.ylabel('Sales')
plt.xlabel('')
```

```python
# to show trend overtime line chart is best fit whereas it is difficult to see trend in bar chart overtime
```

```python
# "Show the distribution of order-line sales values." (Use the raw sales column directly — no summarizing needed.
ax = sns.histplot(data=df,x='sales',bins=10)
plt.title('Distribution of Order-Line Sales Values')
plt.xlabel('Sales')
plt.ylabel('Frequency')
# show the value over each bin
ax.bar_label(ax.containers[0])
#
```

```python
# "Show the relationship between discount level and profit." (Use the raw discount and profit columns directly — no summarizing needed.)
sns.scatterplot(data=df,x='discount',y='profit')
plt.title('Relationship between Discount Level and Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
# as our discount is increasing profit is decreasing
```

# Task 3

```python
sales_by_region = df.groupby('region')['sales'].sum().round(2)
sales_by_region_and_category = df.groupby(['region','category'])['sales'].sum().round(2)
total_profit_by_month = df.groupby(df['order_date'].dt.month)['profit'].sum().round(2)
```

```python
#(a) Recolor the region sales chart with a proper categorical palette.
sns.barplot(sales_by_region,palette='Set2')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
#
```

```python
#(b) Recolor the region-by-category data as a heatmap using a sequential palette.
sns.heatmap(sales_by_region_and_category.unstack(),cmap='Blues')
plt.title('Sales by Region and Category')
plt.xlabel('Category')
plt.ylabel('Region')
#
```

```python
#(c) Build a diverging-color chart of monthly profit, since it crosses zero.
sns.barplot(total_profit_by_month,palette='RdYlGn')
plt.title('Profit by Month')
plt.xlabel('Month')
plt.ylabel('Profit')
#
```

# Task 5

```python
# A truncated y-axis on the region sales chart
ax = sns.barplot(sales_by_region)
ax.set_ylim(100000,500000)
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
# we can't differentiate which categories are perfomring good or not
```

```python
# An inverted y-axis on the monthly profit or sales line
ax= sns.lineplot(x=summarized_monthly_sales.index,y=summarized_monthly_sales.values)
ax.invert_yaxis()
# its showing opposite of the orignal trend
```

```python
fig, ax1 = plt.subplots(figsize=(8, 5))

# First plot (Bar chart on ax1)
sns.lineplot(summarized_monthly_sales, ax=ax1, color="skyblue")
ax1.set_ylabel("Revenue ($)", color="blue")
ax1.set_xlabel('Month')
# Create secondary axis sharing the same x-axis
ax2 = ax1.twinx()

# Second plot (Line chart on ax2)
sns.lineplot(
    total_profit_by_month,
    ax=ax2,
    color="coral",
    marker="o",
)
ax2.set_ylabel("Profit ($)", color="coral")

# Display the chart
plt.title("Revenue and Profit by Month")
```
